```python
from chembl_webresource_client.new_client import new_client
activity = new_client.activity

ache_ic50 = activity.filter(
    target_chembl_id="CHEMBL220",
    standard_type="IC50"
)
print(activity)
ache_ic50 = list(ache_ic50)

print("Number of IC50 records:", len(ache_ic50))
```

```python
record = ache_ic50[0]

fields = [
    "molecule_chembl_id",
    "canonical_smiles",
    "standard_type",
    "standard_relation",
    "standard_value",
    "standard_units",
    "pchembl_value",
    "assay_chembl_id",
    "assay_type",
    "target_chembl_id"
]

for field in fields:
    print(f"{field:22s}: {record[field]}")
```

```python
import pandas as pd

columns = [
    "molecule_chembl_id",
    "canonical_smiles",
    "standard_type",
    "standard_relation",
    "standard_value",
    "standard_units",
    "pchembl_value",
    "assay_chembl_id",
    "assay_type",
    "target_chembl_id"
]

df = pd.DataFrame(ache_ic50)[columns]

df.head()
```

```python
df["standard_relation"].value_counts(dropna=False)
```

```python
df["standard_units"].value_counts(dropna=False)
```

```python
df[
    [
        "canonical_smiles",
        "standard_value",
        "standard_units",
        "pchembl_value"
    ]
].isna().sum()
```

```python
df_exact = df[df["standard_relation"] == "="].copy()

print("Raw records:", len(df))
print("Exact IC50 records:", len(df_exact))
```

```python
df_clean = df_exact[
    df_exact["standard_units"] == "nM"
].copy()

print("Exact records:", len(df_exact))
print("Exact + nM records:", len(df_clean))
```

```python
df_clean[
    [
        "canonical_smiles",
        "standard_value",
        "pchembl_value"
    ]
].isna().sum()
```

```python
df_clean = df_clean.dropna(
    subset=["canonical_smiles", "standard_value"]
).copy()
```

```python
df_clean["standard_value"] = pd.to_numeric(
    df_clean["standard_value"],
    errors="coerce"
)

df_clean = df_clean.dropna(subset=["standard_value"])
```

```python
import numpy as np

df_clean["pIC50"] = -np.log10(df_clean["standard_value"] * 1e-9)
```

```python
df_clean[
    ["standard_value", "standard_units", "pIC50", "pchembl_value"]
].head(10)
```

```python
comparison = df_clean.dropna(subset=["pchembl_value"]).copy()

comparison["pchembl_value"] = pd.to_numeric(
    comparison["pchembl_value"]
)

comparison["difference"] = (
    comparison["pIC50"] - comparison["pchembl_value"]
)

comparison[
    ["standard_value", "pIC50", "pchembl_value", "difference"]
].head(10)
```

```python
comparison["difference"].describe()
```

```python
print("Number of activity records:", len(df_clean))
print(
    "Number of unique molecules:",
    df_clean["molecule_chembl_id"].nunique()
)
```

```python
measurement_counts = (
    df_clean["molecule_chembl_id"]
    .value_counts()
)

measurement_counts.head(10)
```

```python
most_measured = measurement_counts.index[0]

print("Molecule:", most_measured)

df_clean[
    df_clean["molecule_chembl_id"] == most_measured
][
    [
        "molecule_chembl_id",
        "standard_value",
        "standard_units",
        "pIC50",
        "assay_chembl_id",
        "assay_type"
    ]
].sort_values("pIC50")
```

```python
repeat_stats = (
    df_clean
    .groupby("molecule_chembl_id")["pIC50"]
    .agg(["count", "mean", "median", "std", "min", "max"])
)

repeat_stats["range"] = (
    repeat_stats["max"] - repeat_stats["min"]
)

repeat_stats.sort_values(
    "range",
    ascending=False
).head(10)
```

```python
df_final = (
    df_clean
    .groupby("molecule_chembl_id")
    .agg(
        canonical_smiles=("canonical_smiles", "first"),
        pIC50=("pIC50", "median"),
        n_measurements=("pIC50", "count")
    )
    .reset_index()
)

df_final.head()
```

```python
print("Final number of molecules:", len(df_final))
```

```python
from rdkit import Chem

df_final["mol"] = df_final["canonical_smiles"].apply(
    Chem.MolFromSmiles
)

invalid = df_final["mol"].isna().sum()

print("Invalid SMILES:", invalid)
```
